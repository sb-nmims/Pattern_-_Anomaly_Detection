#FACE Regonition

```python
import cv2
```

```python
imagepath = './faces_2.jpg'
cascPath = './haarcascade_frontalcatface.xml'
```

```python
#Read Image
image = cv2.imread(imagepath,0)
```

```python
#create a haar cascade
face = cv2.CascadeClassifier(cascPath)
```

```python
myFace = face.detectMultiScale(image,
                              minSize = (10,10),
                              scaleFactor=1.1,
                              minNeighbors=3,
                              flags=cv2.CASCADE_SCALE_IMAGE)
```

```python
myFace
```

```python
len(myFace)
```

```python
for (x,y,w,h) in myFace:
    cv2.rectangle(image,(x,y),(x+w,y+h),(0,37,255),2)
```

```python
cv2.imshow("",image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
