```python
def average(x):
  try:
    if len(x.split('-')) == 2:
      return (float(x.split('-')[0]) + float(x.split('-')[1]))/2
    else:
      return float(x)
  except:
    return None
```

```python
df['total_sqft'] = df['total_sqft'].apply(average)
```

```python
prop_count = df2['location'].value_counts()
```

```python
more_then_20 = prop_count[prop_count > 20]
```

```python
less_then_20 = prop_count[prop_count <= 20]
```

```python
def location_update(x):
  if x in more_then_20:
    return x
  else:
    'Others'
```

```python
df2['location'] = df2['location'].apply(location_update)
```

```python
df7 = pd.DataFrame()
for key,dfloc in df6.groupby('location'):
  m = dfloc['PPS'].mean()
  s = dfloc['PPS'].std()

  extracted_df = dfloc[(dfloc['PPS'] > (m-2*s)) & (dfloc['PPS'] < (m+2*s))]
  df7 = pd.concat([df7,extracted_df],ignore_index=True)
```

```python
df7.head()
```

```python
df8 = df7.drop(['location','PPS'],axis=1)
```

```python
df8
```

```python
X = df8.drop('price',axis=1)
Y = df8[['price']]
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest = train_test_split(X,Y,train_size=0.8)
```

```python
from sklearn.linear_model import LinearRegression
lmodel = LinearRegression()

lmodel.fit(xtrain,ytrain)
```

```python
ytrain_pred = lmodel.predict(xtrain)
ytest_pred = lmodel.predict(xtest)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error
print(mean_absolute_error(ytrain,ytrain_pred))
print(mean_absolute_error(ytest,ytest_pred))
```

```python
import numpy as np
def predict_price(size,sqft,bath,balcony,location):
  input = np.zeros(len(xtrain.columns))
  input[0] = size
  input[1] = sqft
  input[2] = bath
  input[3] = balcony

  loc = np.where(xtrain.columns == location)
  input[loc] = 1

  return lmodel.predict([input])
```

```python
print(predict_price(3,1750,2,4,'Whitefield'))
```
