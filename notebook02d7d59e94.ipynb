```python
airplane = '../input/natural-images/natural_images/airplane'
car = '../input/natural-images/natural_images/car'
cat = '../input/natural-images/natural_images/cat'
dog = '../input/natural-images/natural_images/dog'
```

```python
from tqdm import tqdm
features = []
for img in tqdm(os.listdir(airplane)):
    iar = cv2.imread(os.path.join(airplane,img),0)
    iar = cv2.resize(iar,(50,50))
    features.append(iar)
    
for img in tqdm(os.listdir(car)):
    iar = cv2.imread(os.path.join(car,img),0)
    iar = cv2.resize(iar,(50,50))
    features.append(iar)
    
for img in tqdm(os.listdir(cat)):
    iar = cv2.imread(os.path.join(cat,img),0)
    iar = cv2.resize(iar,(50,50))
    features.append(iar)
    
for img in tqdm(os.listdir(dog)):
    iar = cv2.imread(os.path.join(dog,img),0)
    iar = cv2.resize(iar,(50,50))
    features.append(iar)
```

```python
import numpy as np
X = np.array(features)
```

```python
X = X.reshape(3282,2500)
```

```python
from tqdm import tqdm
labels = []
for img in tqdm(os.listdir(airplane)):
    labels.append(0)
    
for img in tqdm(os.listdir(car)):
    labels.append(1)
    
for img in tqdm(os.listdir(cat)):
    labels.append(2)
    
for img in tqdm(os.listdir(dog)):
    labels.append(3)
```

```python
Y = np.array(labels)
Y.shape
```

```python
#Using Random forest classifier to create a predictive model
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest = train_test_split(X,Y)
```

```python
from sklearn.ensemble import RandomForestClassifier
rmodel = RandomForestClassifier(max_depth=7)

rmodel.fit(xtrain,ytrain)
```

```python
print(rmodel.score(xtrain,ytrain))
print(rmodel.score(xtest,ytest))
```
