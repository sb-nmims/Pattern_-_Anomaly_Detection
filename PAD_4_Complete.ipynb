```python
#Defining centeroids
import numpy as np
cent = {0 : [np.random.randint(12,72),np.random.randint(7,70)],
        1 : [np.random.randint(12,72),np.random.randint(7,70)],
        2 : [np.random.randint(12,72),np.random.randint(7,70)]}
cent
```

```python
import matplotlib.pyplot as plt

plt.scatter(data['x'],data['y'])

plt.scatter(cent[0][0],cent[0][1])
plt.scatter(cent[1][0],cent[1][1])
plt.scatter(cent[2][0],cent[2][1])

plt.show()
```

```python
#Calculating euclidean distances from each centeroids
data['0'] = ((data['x'] - cent[0][0])**2 + (data['y'] - cent[0][1])**2)**0.5
data['1'] = ((data['x'] - cent[1][0])**2 + (data['y'] - cent[1][1])**2)**0.5
data['2'] = ((data['x'] - cent[2][0])**2 + (data['y'] - cent[2][1])**2)**0.5
```

```python
#Assigning values to each cluster
data['min_dist.'] = data.loc[:,'0':'2'].min(axis=1)
data['min_dist._id'] = data.loc[:,'0':'2'].idxmin(axis=1)

data['color'] = data['min_dist._id'].map({'0':'r' , '1':'g' , '2':'b'})
```

```python
import matplotlib.pyplot as plt

plt.scatter(data['x'],data['y'],c=data['color'])

plt.show()
```

```python
#Update the centeroids by calculating crossponding means
cent[0][0] = data[data['min_dist._id'] == '0']['x'].mean()
cent[0][1] = data[data['min_dist._id'] == '0']['y'].mean()

cent[1][0] = data[data['min_dist._id'] == '1']['x'].mean()
cent[1][1] = data[data['min_dist._id'] == '1']['y'].mean()

cent[2][0] = data[data['min_dist._id'] == '2']['x'].mean()
cent[2][1] = data[data['min_dist._id'] == '2']['y'].mean()
```

#Mall data for clustering

```python
import pandas as pd
dfm = pd.read_csv('mall.csv')
dfm.head()
```

```python
dfX = dfm[['Annual Income (k$)','Spending Score (1-100)']]
dfX.columns = ['Ai','Sc']
dfX.head()
```

```python
plt.scatter(dfX['Ai'],dfX['Sc'])
plt.show()
```

```python
import numpy as np

def ini_centeroids(x):
  cent = {}
  for i in range(x):
    cent[i] = [np.random.randint(dfX['Ai'].min() , dfX['Ai'].max()) ,
               np.random.randint(dfX['Sc'].min() , dfX['Sc'].max())]
  return cent
```

```python
def assign_cluster(x):
  for i in range(x):
    dfX[str(i)] = ((dfX['Ai'] - cent[i][0])**2 + (dfX['Sc'] - cent[i][1])**2)**0.5

  dfX['min_dist._id'] = dfX.loc[:,'0':str(x-1)].idxmin(axis=1)
  dfX['min_dist._id'] = dfX['min_dist._id'].astype(int)
```

```python
def update_centeroids(x):
  for i in range(x):
    cent[i][0] = dfX[dfX['min_dist._id'] == i]['Ai'].mean()
    cent[i][1] = dfX[dfX['min_dist._id'] == i]['Sc'].mean()
```

```python
cent = ini_centeroids(5)
cent
```

```python
assign_cluster(5)
```

```python
for i in range(300):
  update_centeroids(5)
  assign_cluster(5)
```

#More then two features for clustring, Library to be used, Elbow rule

```python
import pandas as pd
dfm = pd.read_csv('mall.csv')
dfm.head()
```

```python
df = dfm[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']]
```

```python
df.head()
```

```python
from sklearn.cluster import KMeans
kmodel = KMeans(n_clusters=5)

kmodel.fit(df)

df['Label'] = kmodel.predict(df)
```

```python
df.head()
```

```python
df.groupby('Label')[['Age','Annual Income (k$)','Spending Score (1-100)']].mean()
```

#Knee rule or elbow method to decide number of clusters

```python
losses = []
for i in range(1,14):
  kmeans_model = KMeans(n_clusters=i)
  kmeans_model.fit(df)
  losses.append(kmeans_model.inertia_)
```

```python
plt.plot(range(1,14),losses)
plt.show()
```
