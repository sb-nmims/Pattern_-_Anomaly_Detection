```python
import pandas as pd
df = pd.read_csv('heart.csv')
df.head()
```

```python
df['DEATH_EVENT'].value_counts()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
X = df.drop('DEATH_EVENT',axis=1).values
Y = df[['DEATH_EVENT']].values
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest = train_test_split(X,Y)
```

```python
from sklearn.svm import SVC
smodel1 = SVC(C=1.0, kernel='linear')

smodel1.fit(xtrain,ytrain)
```

```python
print(smodel1.score(xtrain,ytrain))
print(smodel1.score(xtest,ytest))
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(ytest,smodel1.predict(xtest))
```

#Polynomial SVM Kernel

```python
from sklearn.svm import SVC
smodel2 = SVC(C=2.0, kernel='poly' , degree=4)

smodel2.fit(xtrain,ytrain)
```

```python
print(smodel2.score(xtrain,ytrain))
print(smodel2.score(xtest,ytest))
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(ytest,smodel2.predict(xtest))
```

#RBF kernel

```python
from sklearn.svm import SVC
smodel3 = SVC(C=2.0, kernel='rbf',degree=5)

smodel3.fit(xtrain,ytrain)
```

```python
print(smodel3.score(xtrain,ytrain))
print(smodel3.score(xtest,ytest))
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(ytest,smodel3.predict(xtest))
```

#On Applying SVM on the following data find out which kernel gives best accurate model

```python
from sklearn.datasets import load_breast_cancer
df = load_breast_cancer()
```

```python
df.keys()
```

```python
import pandas as pd
X = pd.DataFrame(df['data'],columns=df['feature_names'])
Y = df['target']
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest = train_test_split(X,Y)
```

```python
from sklearn.svm import SVC
smodel_c = SVC(C=1.0, kernel='linear')

smodel_c.fit(xtrain,ytrain)
```

```python
print(smodel_c.score(xtrain,ytrain))
print(smodel_c.score(xtest,ytest))
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(ytest,smodel_c.predict(xtest))
```
