```python
import pandas as pd
df = pd.read_csv('/content/Advertising.csv',index_col=0)
df.head()
```

```python
X = df[['TV','radio']].values
Y = df[['sales']].values
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest = train_test_split(X,Y)
```

```python
from sklearn.preprocessing import PolynomialFeatures
pol = PolynomialFeatures(2,include_bias=False)
```

```python
pol.fit(xtrain)
```

```python
xtrain_pol = pol.transform(xtrain)
xtest_pol = pol.transform(xtest)
```

```python
from sklearn.linear_model import LinearRegression
lmodel = LinearRegression()

lmodel.fit(xtrain_pol,ytrain)
```

```python
print(lmodel.coef_)
print(lmodel.intercept_)
```

```python
ytrain_pred = lmodel.predict(xtrain_pol)
ytest_pred = lmodel.predict(xtest_pol)
from sklearn.metrics import mean_absolute_error
print(mean_absolute_error(ytrain_pred,ytrain))
print(mean_absolute_error(ytest_pred,ytest))
```

```python
import numpy as np
np.hstack((ytest,ytest_pred))
```

```python
#to predict - TV - 54, Radio - 89
uv = [[54 , 89]]
print(lmodel.predict(pol.transform(uv))[0][0])
```

```python
def predict_sales(x1,x2):
  pol_f = pol.transform([[x1,x2]])
  return lmodel.predict(pol_f)
```
