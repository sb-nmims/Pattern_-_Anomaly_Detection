```python
from math import exp
coef = [0.55,0.45,1.10]

yhat = coef[0]
yhat += coef[1] * data[:,0:1]
yhat += coef[2] * data[:,1:2]
```

```python
for val in yhat:
  z = 1.0/(1.0 + exp(-val))
  print(z,round(z))
```

```python
import numpy as np
data = np.array([[3.87,2.65,0],
                 [2.64,3.63,0],
                 [4.93,5.50,0],
                 [2.83,2.95,0],
                 [4.07,4.00,0],
                 [8.27,3.57,1],
                 [6.36,3.09,1],
                 [7.99,2.97,1],
                 [9.76,1.42,1],
                 [8.76,4.60,1]])
print(data)
```

```python
def predict(coef,row):
  yhat = coef[0]
  for i in range(2):
    yhat += coef[i+1] * row[i]
  return 1.0/(1.0 + exp(-yhat))
```

```python
coef = [0.55,0.45,1.10]
for row in data:
  z = predict(coef,row)
  print('expected',row[2], 'Predicted',z , 'final',round(z))
```

```python
lr = 0.3
for epoch in range(30):
  sum_error = 0
  for row in data:
    z = predict(coef,row)
    error = row[-1] - z
    sum_error += error**2
    coef[0] = coef[0] + lr * error * z * (1-z)
    for i in range(2):
      coef[i+1] = coef[i+1] + lr * error * z * (1-z) * row[i]

  print(epoch,sum_error)
```

```python
print(coef)
```

```python
for row in data:
  z = predict(coef,row)
  print('expected',row[2], 'Predicted',z , 'final',round(z))
```

#Logistic Regression Using Library

```python
import numpy as np
data = np.array([[3.87,2.65,0],
                 [2.64,3.63,0],
                 [4.93,5.50,0],
                 [2.83,2.95,0],
                 [4.07,4.00,0],
                 [8.27,3.57,1],
                 [6.36,3.09,1],
                 [7.99,2.97,1],
                 [9.76,1.42,1],
                 [8.76,4.60,1]])
print(data)
```

```python
X = data[:,0:2]
Y = data[:,2:]
```

```python
from sklearn.linear_model import LogisticRegression
lmodel = LogisticRegression()

lmodel.fit(X,Y)
```

```python
lmodel.predict(X)
```

```python
lmodel.predict_proba(X)
```
