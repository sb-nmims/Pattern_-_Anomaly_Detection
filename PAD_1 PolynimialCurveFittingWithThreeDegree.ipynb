**Polynomial curve fitting**

```python
import numpy as np
x = np.array([3,5,6.1,7,9])  #independent variable
y = np.array([4.1,5,7.2,6.9,8])   #Dependent variable
```

```python
import matplotlib.pyplot as plt

plt.scatter(x,y)

plt.show()
```

```python
x = x.reshape(5,1)
y = y.reshape(5,1)
```

```python
X = np.hstack((x,x**2,x**3,x**4))
X
```

```python
a = 2.3
b = 1.5
c = 1.7
d = 2.1
e = 3.2
Yh = a + b*X[:,0:1] + c*X[:,1:2] + d*X[:,2:3] + e*X[:,2:3]
Yh
```

```python
error = ((y - Yh)**2).mean()
print(error)
```

```python
dea =  -2*((y-Yh)).sum()
deb = -2*((y-Yh)*X[:,0:1]).sum()
dec = -2*((y-Yh)*X[:,1:2]).sum()
ded = -2*((y-Yh)*X[:,2:3]).sum()
dee = -2*((y-Yh)*X[:,3:4]).sum()

print(dea,deb,dec,ded)
```

```python
lr = 1e-7
a = a - lr*dea
b = b - lr*deb
c = c - lr*dec
d = d - lr*ded
e = e - lr*dee
```

```python
Yh = a + b*X[:,0:1] + c*X[:,1:2] + d*X[:,2:3] + e*X[:,2:3]
error = ((y - Yh)**2).mean()
print(error)
```

```python
for i in range(1000):
  Yh = a + b*X[:,0:1] + c*X[:,1:2] + d*X[:,2:3] + e*X[:,2:3]

  dea =  -2*((y-Yh)).sum()
  deb = -2*((y-Yh)*X[:,0:1]).sum()
  dec = -2*((y-Yh)*X[:,1:2]).sum()
  ded = -2*((y-Yh)*X[:,2:3]).sum()
  dee = -2*((y-Yh)*X[:,3:4]).sum()

  lr = 1e-7
  a = a - lr*dea
  b = b - lr*deb
  c = c - lr*dec
  d = d - lr*ded
  e = e - lr*dee

  error = ((y - Yh)**2).mean()
  print(i,error)
```

```python
import matplotlib.pyplot as plt

plt.scatter(x,y)
plt.plot(x,Yh,c='r')

plt.show()
```
