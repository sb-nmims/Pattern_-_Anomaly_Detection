```python
import numpy as np
from numpy.linalg import eig
```

```python
A = np.array([[6,5],[2,3],[1,4]])
A
```

```python
M = np.mean(A,axis=0)
M
```

```python
#Standardise the values
C = A - M
print(C)
```

```python
#Calculate covariance matrix of each feature
V = np.cov(C.T)
V
```

```python
#Eigen vectors of covariance matrix
values,vectors = eig(V)
print(values)
```

```python
print(vectors)
```

```python
#Project the data
P = np.dot(C,vectors)
```
