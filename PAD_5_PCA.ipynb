```python
import numpy as np
A = np.array([[6,5,6,8,9] , [2,3,7,7,2] , [1,4,3,9,0] , [1,4,3,9,0]])
A
```

```python
from sklearn.decomposition import PCA
pca = PCA(3)
```

```python
pca.fit(A)
```

```python
pca.transform(A)
```

```python
pca.explained_variance_
```

```python
pca.explained_variance_ratio_
```

#PCA on real Data

```python
from sklearn.datasets import load_breast_cancer
df = load_breast_cancer()
```

```python
df.keys()
```

```python
df['target_names']
```

```python
import pandas as pd
X = pd.DataFrame(df['data'],columns=df['feature_names'])
```

```python
X.shape
```

```python
X
```

```python
from sklearn.decomposition import PCA
pca = PCA(30)
```

```python
pca.fit(X)
X_pca = pca.transform(X)
```

```python
X_pca.shape
```

```python
pd.DataFrame(X_pca)
```

```python
pca.explained_variance_
```

```python
import matplotlib.pyplot as plt
plt.scatter(X_pca[:,0:1] , X_pca[:,1:2] , c=df['target'])
plt.show()
```
