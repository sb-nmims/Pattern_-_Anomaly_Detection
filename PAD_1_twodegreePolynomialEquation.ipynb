```python
import numpy as np
x = np.array([3,5,6.1,7,9])  #independent variable
y = np.array([4.1,5,7.2,6.9,8])   #Dependent variable
```

```python
a = 7.1
b = 8.7
yh = a*x + b
error = ((y - yh)**2).mean()
print(error)
```

```python
lr = 0.001
for i in range(100):
  yh = a*x + b

  dea = -2*((y-yh)*x).sum()
  deb = -2*((y-yh)).sum()

  a = a - lr * dea
  b = b - lr * deb

  error = ((y - yh)**2).mean()
  print(i,error)
```
